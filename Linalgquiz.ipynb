```python
#import required packages
import numpy as np
```

```python
M = np.array([[1.0,  -4.0, -0.5, -2.5,  2.0],
              [-4.0,  3.0, -5.5, -5.5, -3.5],
              [-0.5, -5.5,  7.0, -2.0,  3.5],
              [-2.5, -5.5, -2.0, -2.0, -4.0],
              [ 2.0, -3.5, -3.5, -4.0,  1.0]])

b = np.array([1, -1, 2, -1, 0])
```

```python
for i in range(M.shape[0]):
    terms = []
    for j in range(M.shape[1]):
        terms.append("{1} x[{0}]".format(j, M[i, j]))
    print(" + ".join(terms), "=", b[i])
```

```python
def qrdecomp(M):
  n=M.shape[0]
  M_working=np.copy(M)
  Q=np.zeros((n,n))
  R=np.zeros((n,n))
  for i in range(n):
    for j in range(i):
      R[j,i]=np.sum(Q[:,j]*M[:,i])
      M_working[:,i]-=R[j,i]*Q[:,j]
  R[i,i]=np.sqrt(np.sum(M_working[:,i]*M_working[:,i]))
  Q[:,i]=M_working[:,i]/R[i,i]

  return Q,R


print("Matrix Q:")
print(Q)
print("\nMatrix R:")
print(R)
```

```python
Q,R=  np.linalg.qr(M) #run as doubvle check

print("Matrix Q:")
print(Q)
print("\nMatrix R:")
print(R)
```

```python
print("Result of Q^T * Q: ")
print(Q.T @ Q)
```

```python
y = Q.T@b

x =  np.linalg.inv(R) @ y

print("The solution to the provided system of equations is: ", x)
```

```python
x_M = np.linalg.solve(M, b)

print("The solution to the provided system of equations is: ", x_M)
```

```python
A = np.array([[16, 10, 7],
              [10, -8, 0],
              [7, 0, 6]])
```

```python
def qr_eigen (input,itmax=100):
  A_copy=np.copy(A)
  for iteration in range(1,itmax):
    Q,R=qrdecomp(A_copy)
    A_copy=np.dot(R,Q)

  eigvalues=np.diag(A_copy)
  eigvectors=np.linalg.inv(Q)
  return eigvalues,eigvectors


print(f"\nEigenvalues: {eigvalues}")
print(f"Eigenvectors:\n{eigvectors}")
```

```python
eigvalues, eigvectors = np.linalg.eig(A)#check against np

print(f"\nEigenvalues: {eigvalues}")
print(f"Eigenvectors:\n{eigvectors}")
```

```python
D = np.diag(eigvalues)

P = eigvectors

print("Matrix D:")
print(D)

print("\nMatrix P:")
print(P)
```

```python
P_inverse = np.linalg.inv(P)

print("Matrix P-inverse:")
print(P_inverse)
```
